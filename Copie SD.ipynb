```python
import os
import datetime
import shutil


# Trouver la date courante
today = datetime.date.today()

# Enregistrer date dans variable
date_of_day = today.strftime('%y %m ')

# Répertoires source et destination
source_dir = "E:/DCIM/100EOS7D"
X=str(input("Saisir texte :"))
Slash = "/"
destination = "D:/# Mes Pellicules photo/"
destination_dir = destination + date_of_day + X + Slash

# Création nouveau répertoire
os.mkdir(destination_dir)

# Liste de tous les fichiers
files = os.listdir(source_dir)

# Recherche .CR3
for file in files:
    if file.endswith(".CR3") :

        file_path = os.path.join(source_dir, file)
        dest_file_path = os.path.join(destination_dir, file)

        # Copier les fichier
        shutil.copy(file_path, dest_file_path)
```
